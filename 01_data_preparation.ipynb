## Limpieza de valores nulos

Tras el análisis exploratorio realizado en la fase de Data Understanding, se ha detectado la presencia de valores nulos en variables clave, por lo que se ha diseñado una estrategia de imputación diferenciada según su naturaleza para evitar la pérdida de registros.

Para las variables categóricas como region_code, entry_channel y segment, los valores faltantes se agruparán bajo la etiqueta 'Other', evitando así sesgar el modelo hacia categorías específicas; por su parte, en las variables gender, payroll y pension_plan se aplicará una imputación por moda al ser el valor más frecuente y representativo de la muestra.

Finalmente, para la variable numérica salary, se optará por la mediana como medida de tendencia central, garantizando una imputación robusta frente a los valores extremos detectados y manteniendo la integridad técnica de la columna para el posterior modelado.

```python
for name, df in dfs.items():
  nulls_count = df.isnull().sum()
  if nulls_count.sum()>0:
    print(f'\n Dataset {name}')
    print(nulls_count[nulls_count>0])
```

```python
def universal_cleaner(df):
    # 1. Definimos los grupos de variables según la estrategia
    to_other = ['region_code', 'entry_channel', 'segment']
    to_moda = ['gender', 'payroll', 'pension_plan']
    to_median = ['salary']

    # 2. Aplicamos la lógica solo si la columna existe en el dataset
    for col in df.columns:
        if col in to_other:
            df[col] = df[col].fillna('Other')

        elif col in to_moda:
            # Solo si hay nulos calculamos la moda para evitar errores
            if df[col].isnull().any():
                df[col] = df[col].fillna(df[col].mode()[0])

        elif col in to_median:
            df[col] = df[col].fillna(df[col].median())

    return df

# 3. LLAMADA A LA FUNCIÓN
df_sales = universal_cleaner(df_sales)
df_product_desc = universal_cleaner(df_product_desc)
df_customdemo = universal_cleaner(df_customdemo)
df_custom_prod = universal_cleaner(df_custom_prod)
df_customcommact = universal_cleaner(df_customcommact)

# 4. COMPROBACIÓN FINAL
print("Comprobación de nulos tras la limpieza:")
print(f"- Salary en customdemo: {df_customdemo['salary'].isnull().sum()}")
print(f"- Entry Channel en customcommact: {df_customcommact['entry_channel'].isnull().sum()}")
print(f"- Gender en customdemo: {df_customdemo['gender'].isnull().sum()}")
```

```python
# Convertimos 'pk_partition' (ej: '2019-02') a un periodo mensual 'YYYY-MM'
# Esto facilita que Pandas lo entienda como tiempo y no como un simple texto
df_merge['month_key'] = pd.to_datetime(df_merge['pk_partition']).dt.to_period('M')

# Para 'month_sale' (ej: '2019-02-01'), al convertirlo a periodo 'M',
# Pandas automáticamente ignora el '-01' y lo alinea con las otras tablas
df_sales['month_key'] = pd.to_datetime(df_sales['month_sale']).dt.to_period('M')

# -------------------------------------------------------------------
# AGREGACIÓN DE VENTAS (EVITAR DUPLICADOS)
# -------------------------------------------------------------------

# Sumamos el margen neto de todas las ventas que un cliente hizo en un mismo mes
sales_monthly = df_sales.groupby(['pk_cid', 'month_key'])['net_margin'].sum().reset_index()

# -------------------------------------------------------------------
# EL MERGE MAESTRO (AÑADIR EL TARGET / RENTABILIDAD)
# -------------------------------------------------------------------

# Unimos con las ventas usando 'left join'.
# Los clientes que NO compraron ese mes tendrán un margen de NaN.
df_merge = pd.merge(df_merge, sales_monthly,
                    on=['pk_cid', 'month_key'],
                    how='left')

# -------------------------------------------------------------------
# LIMPIEZA FINAL
# -------------------------------------------------------------------
# Rellenamos los NaN de 'net_margin' con 0 (lógico: no hubo compra, no hay margen)
df_merge['net_margin'] = df_merge['net_margin'].fillna(0)

# Borramos la columna auxiliar 'month_key' para no ocupar RAM innecesariamente
df_merge.drop(columns=['month_key'], inplace=True, errors='ignore')

print(f"Merge exitoso. Filas de la tabla final: {df_merge.shape[0]}")
```

Una vez consolidado el dataset final (df_reduced), se ha procedido a categorizar las variables en grupos lógicos según su naturaleza y el tratamiento técnico que requieren. Esta segmentación mediante listas permite automatizar la aplicación de estrategias de transformación específicas —como el desglose de fechas, la codificación de variables categóricas o la normalización de valores numéricos— garantizando un preprocesamiento estructurado y eficiente para el modelo de Machine Learning.

```python

def auto_classify_columns(df):
    """
    Clasifica automáticamente las columnas del dataframe según su contenido
    y tipo de dato
    """

    # 1. FECHAS: Identifica columnas que contienen 'date', 'partition' o 'month' en el nombre
    list_dates = [col for col in df.columns if any(x in col.lower() for x in ['date', 'partition', 'month_sale'])]

    # 2. BINARIAS/PRODUCTOS: Columnas con solo 2 valores únicos
    # Excluimos las que ya son fechas
    list_binary = [col for col in df.columns if df[col].nunique() == 2 and col not in list_dates]

    # 3. NUMÉRICAS: Columnas con tipo de dato número (float/int) que no sean binarias ni fechas
    list_numerical = [col for col in df.columns if pd.api.types.is_numeric_dtype(df[col])
                      and col not in list_binary and col not in list_dates]

    # 4. CATEGÓRICAS:
    list_categorical = [col for col in df.columns if col not in list_dates
                        and col not in list_binary and col not in list_numerical]

    return list_dates, list_binary, list_numerical, list_categorical

# Uso de la función
dates, bins, nums, cats = auto_classify_columns(df_reduced)

# Verificación rápida
print(f"Fechas: {dates}")
print(f"Binarias: {bins}")
print(f"Numéricas: {nums}")
print(f"Categorias: {cats}")
```

### DATA PREPARATION FECHAS

El tratamiento de las variables temporales se ha centrado en la descomposición de las fechas en componentes numéricos procesables por los algoritmos. Dada la naturaleza de los datos bancarios, se detectó una excepción crítica: la presencia de registros con etiquetas de texto como 'sin_venta' y valores nulos. Posteriormente, se extrajeron el año y el mes en columnas independientes, eliminando la columna original. Para garantizar que el modelo pudiera interpretar los casos sin fecha sin perder la fila, se imputaron estos valores con el valor constante -1.

```python
df_dates = df_reduced.copy()
```

```python
for col in dates:
    # Convertimos a datetime (esto detecta los 'sin_venta' y los hace NaT)
    temp_dt = pd.to_datetime(df_dates[col], errors='coerce')

    # Extraemos Año y Mes con tipos de datos pequeños
    df_dates[f'{col}_year'] = temp_dt.dt.year.fillna(-1).astype('int16')
    df_dates[f'{col}_month'] = temp_dt.dt.month.fillna(-1).astype('int8')

    # Eliminamos la columna original inmediatamente para liberar espacio
    df_dates.drop(columns=[col], inplace=True)

    # Liberamos memoria
    del temp_dt

# 3. Limpieza final de memoria
import gc
gc.collect()

# 4. Comprobación rápida (solo las nuevas columnas)
print("Columnas de fecha procesadas:")
print(df_dates.filter(like='_year').columns.tolist())
print(df_dates.filter(like='_month').columns.tolist())
```

### DATA PREPARATION NUMERICAS

Para las variables numéricas continuas, como el salario y la edad, el proceso se enfocó en mitigar el impacto de los valores atípicos (outliers) y simplificar la escala de los datos. Se identificó una excepción relevante en las distribuciones de ingresos y edades, con presencia de valores extremadamente altos y registros incoherentes como edades mínimas de 2 años,

Si bien estas edades podrían corresponder a cuentas de ahorro abiertas por tutores legales para el futuro de menores, se determinó que estos perfiles no son el público objetivo para el crecimiento de la cartera de productos activos.

Para corregirlo, se aplicó una técnica de capping, limitando los salarios al percentil 95 y restringiendo el rango de edad entre los 18 y 90 años. Una vez controlados los extremos, los valores faltantes se imputaron con la mediana de cada variable.

```python
df_num = df_dates.copy()
```

```python
for col in nums:
    # A. Conversión inicial
    temp_series = pd.to_numeric(df_num[col], errors='coerce')

    # B. Limpieza específica de OUTLIERS
    if 'salary' in col.lower():
        # Capamos salarios: todo lo que supere el percentil 95
        limite_superior = temp_series.quantile(0.95)
        temp_series = temp_series.clip(upper=limite_superior)

    elif 'age' in col.lower():
        # Capamos edades: entre 18 y 90 años
        # Los menores de 18 o mayores de 90 se ajustan a esos límites
        temp_series = temp_series.clip(lower=18, upper=90)

    temp_series = temp_series.round(2)

    # C. Imputación con la mediana del valor ya "limpio"
    mediana = temp_series.median()
    df_num[f'{col}_n'] = temp_series.fillna(mediana).astype('float32')

    # D. Eliminamos original
    df_num.drop(columns=[col], inplace=True)
    del temp_series

# 3. Limpieza de memoria
if 'df_dates' in locals():
    del df_dates
import gc
gc.collect()
```

### DATA PREPARATION CATEGORICAL

En cuanto a las variables cualitativas, el principal desafío fue la alta cardinalidad de ciertas dimensiones, lo que amenazaba con generar una matriz de datos excesivamente dispersa tras el proceso de codificación.

La estrategia consistió en una reducción de dimensionalidad selectiva: para cada variable, se preservaron únicamente los cinco niveles con mayor frecuencia de aparición.

Todos los valores restantes, considerados minoritarios o excepcionales, se reclasificaron bajo una etiqueta unificada denominada 'Otros'.

Una vez simplificadas, se transformaron mediante One-Hot Encoding para generar nuevas columnas binarias, eliminando sistemáticamente las variables originales.


```python
df_final = df_num.copy()

# Procesamos cada columna categórica
for col in cats:
    print(f"Transformando: {col}")

    # Identificamos los 5 valores más frecuentes
    top_7 = df_final[col].value_counts().nlargest(7).index.tolist()

    #  Creamos una versión temporal con los valores agrupados
    # Los que no son Top 5 pasan a ser 'Otros'
    temp_grouped = df_final[col].where(df_final[col].isin(top_7), 'Otros')

    # Generamos las columnas One-Hot Encoding (Dummies)
    # Usamos dtype='int8' para que no pese casi nada
    dummies = pd.get_dummies(temp_grouped, prefix=f"{col}_cat", drop_first=False, dtype='int8')

    # CONCATENAMOS las nuevas y ELIMINAMOS la vieja
    df_final = pd.concat([df_final, dummies], axis=1)
    df_final.drop(columns=[col], inplace=True)

    # Limpieza inmediata de RAM
    del temp_grouped
    del dummies
    gc.collect()

# 3. Borramos el dataset anterior
if 'df_num' in locals():
    del df_num

for ds in dfs:
    if ds in locals():
        del locals()[ds]



gc.collect()
```

### Data Preparation Boolean

La fase final de la preparación consistió en la homogeneización de todas las variables binarias y de estado al tipo de dato lógico boolean.

En este bloque se gestionó la excepción de la variable gender, que originalmente se presentaba como una cadena de texto; esta se mapeó a valores numerico (1/0) para integrarla eficientemente con el resto del dataset.

De igual manera, las variables de tenencia de productos se transformaron desde su formato numérico a booleano, interpretando cualquier valor nulo como una ausencia de contratación.


```python
if 'gender' in df_final.columns:
  '''
  Tal y como se observo en el data understanding, el atributo Gender es un atributo
  en formato de string, po lo que es necesario convertirlo en un formato numerico
  o booleano

  '''
df_final['gender_b'] = df_final['gender'].map({'V': 1, 'H': 0}).fillna(False).astype('int8')

# Borramos la original
df_final.drop(columns=['gender'], inplace=True)

# 2. Procesamos el resto de las binarias
for col in bins:
    if col in df_final.columns: # Por si acaso alguna ya se borró
        # Creamos la nueva booleana
        df_final[f'{col}_b'] = pd.to_numeric(df_final[col], errors='coerce').fillna(0).astype('int8')

        # Eliminamos la original
        df_final.drop(columns=[col], inplace=True)

gc.collect()
```
